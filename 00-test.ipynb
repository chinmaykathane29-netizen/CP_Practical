```python
# gets marks of test student based on the marks
#above 95 - grade A , above 85 - grade B , above 75 - grade C , above 65 - grade D , above 55 - grade E , below 55 - grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    grade = "A"
elif marks >= 85:
    grade = "B"
elif marks >= 75:
    grade = "C"
elif marks >= 65:
    grade = "D"
elif marks >= 55:
    grade = "E"
else:
    grade = "F"
print("Your grade is: " + grade, "\nand your marks are: ", marks)
```
